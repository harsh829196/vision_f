```python
import os
import zipfile

# Create target data directory
os.makedirs("data", exist_ok=True)
zip_temp = "data/kvasir-seg.zip"

# Download directly using wget, bypassing SSL certificate verification
print("Downloading dataset via wget...")
!wget --no-check-certificate -O {zip_temp} https://datasets.simula.no/downloads/kvasir-seg.zip

# Extract zip to the expected destination
if os.path.exists(zip_temp) and os.path.getsize(zip_temp) > 0:
    print("Extracting dataset...")
    with zipfile.ZipFile(zip_temp, 'r') as zip_ref:
        zip_ref.extractall("data")

    # Clean up zip
    os.remove(zip_temp)

# Verify the downloaded files
image_path = "data/Kvasir-SEG/images"
mask_path = "data/Kvasir-SEG/masks"

if os.path.exists(image_path) and os.path.exists(mask_path):
    print("Extraction successful!")
    print("Images:", len(os.listdir(image_path)))
    print("Masks:", len(os.listdir(mask_path)))
else:
    print("Dataset directories were not found. Please check the download logs above.")
```
